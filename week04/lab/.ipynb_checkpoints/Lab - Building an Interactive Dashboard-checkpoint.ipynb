## 13. Create the first static Plotly figure

Build the first figure before placing it in Dash. Give it a descriptive title and human-readable axis and legend labels.



```python
fig1 = px.bar(
    analysis.groupby('major_category', as_index=False)['earnings_10yr_usd'].mean(),
    x='major_category',
    y='earnings_10yr_usd',
    title='Average 10-Year Earnings by Major Category',
    labels={
        'major_category': 'Major Category',
        'earnings_10yr_usd': 'Average 10-Year Earnings (USD)'
    }
)

fig1.show()
```

**What evidence should a viewer obtain from the first figure?**
The average earnings over a 10-year period from each of the seven different major categories.


## 14. Create the second static Plotly figure

The second figure should add different but related evidence rather than repeat the first figure in another style.



```python
fig2 = px.bar(
    analysis.groupby('major_category', as_index=False)['roi_pct'].mean(),
    x='major_category',
    y='roi_pct',
    title='Average Return on Investment by Major Category',
    labels={
        'major_category': 'Major Category',
        'roi_pct': 'Average ROI (%)'
    }
)

fig2.show()
```

**What does the second figure add to the first figure's evidence?**
It shows the Return on Investment of the seven major categories. It adds information about how the earnings compare to the cost of the program, showing whether each major category has a positive or negative average ROI.

## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
def make_figures(selected_category):
    filtered = analysis[analysis['major_category'] == selected_category]

    summary = (
        filtered.groupby('major', as_index=False)
        .agg(
            earnings_10yr_usd=('earnings_10yr_usd', 'mean'),
            roi_pct=('roi_pct', 'mean')
        )
    )

    fig1 = px.bar(
        summary,
        x='major',
        y='earnings_10yr_usd',
        title=f"Average 10-Year Earnings by Major: {selected_category}",
        labels={
            'major': 'Major',
            'earnings_10yr_usd': 'Average 10-Year Earnings (USD)'
        }
    )

    fig2 = px.bar(
        summary,
        x='major',
        y='roi_pct',
        title=f"Average Return on Investment by Major: {selected_category}",
        labels={
            'major': 'Major',
            'roi_pct': 'Average ROI (%)'
        }
    )
    
    return fig1, fig2

```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1("College Major ROI Dashboard"),

    html.P(
        "Explore how 10-year earnings and return on investment vary "
        "across college major categories."
    ),

    html.Label("Select a major category:"),

    dcc.Dropdown(
        id="major-category-dropdown",
        options=major_category_options,
        value="STEM"
    ),

    dcc.Graph(id='earnings-graph'),

    dcc.Graph(id='roi-graph'),

    html.H3("Conclusion"),
    html.P(
        "The dashboard compares 10-year earnings and ROI across "
        "different college major categories."
    ),

    html.H3("Limitation"),
    html.P(
        "The dataset contains synthetic graduate records, so the "
        "results represent modeled outcomes rather than actual "
        "observed graduate earnings."
    )
])

```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python
@app.callback(
    Output("earnings-graph", "figure"),
    Output("roi-graph", "figure"),
    Input("major-category-dropdown", "value")
)

def update_dashboard(selected_category):
    fig1, fig2 = make_figures(selected_category)
    return fig1, fig2

```

## 19. Check the callback as a Python function

Call the callback or reusable figure function with at least two valid control values. Confirm that each call returns two Plotly figures without an error.



```python
update_dashboard("STEM")

```

```python
update_dashboard("business")
```
